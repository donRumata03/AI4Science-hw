# NCM personal-library search
English adaptation of course search_lab.ipynb cells 5, 9, 13, 19, 34 and 45.
The retrieval run is implemented in checks/run_search_lab.py; these executed cells inspect its saved outputs.
To refresh retrieval, run `.venv/Scripts/python.exe checks/run_search_lab.py` from the repository root.

```
import json
from pathlib import Path
root = Path.cwd()
if not (root / 'library').exists(): root = root.parent
data = json.loads((root / 'library/retrieval/search-results.json').read_text())
print(data['documents'], 'documents;', len(data['results']), 'queries;', data['encoder'])
for row in data['results']:
    print(row['query'])
    for method in ('bm25', 'vector'):
        print(method, [item['title'] for item in row['rankings'][method][:3]])
```
